```python
import matplotlib.pyplot as plt
```

```python
x=[2,3,4,5,6]
y=[5,6,7,8,9]
y1=[4,5,6,7,8]
plt.plot(x,y)
plt.show()
```

```python
plt.subplot(2,2,1)
plt.suptitle("Data")
plt.title("graph 1")
plt.xlabel("x axis")
plt.ylabel("y axis")
plt.plot(x,y)
plt.subplot(2,2,2)
plt.title("graph 2")
plt.xlabel("x axis")
plt.ylabel("y axis")
plt.plot(x,y1)
plt.show()
```

```python
plt.subplot(2,2,3)
plt.suptitle("Data")
plt.title("graph 3")
plt.xlabel("x axis")
plt.ylabel("y axis")
plt.plot(y,x)
plt.subplot(2,2,4)
plt.title("graph 4")
plt.xlabel("x axis")
plt.ylabel("y axis")
plt.plot(y1,y)
plt.show()
```
